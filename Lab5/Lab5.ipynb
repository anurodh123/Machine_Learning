# Lab Session 4: Logistic Regression
## Dataset: Breast Cancer (Scikit-learn)

```
import numpy as np
import pandas as pd
```

## Task 1: Load the Dataset

```
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
x = data.data
y = data.target
```

## Task 2: Explore the Dataset
1. Display the shape of X and y.
2. Print the first five rows.
3. Identify:
    * Number of samples
    * Number of features
    * Number of classes

```
print("=== Task 2: Explore the Dataset ===")
print("Shape of X:", x.shape)
print("Shape of y:", y.shape)

df = pd.DataFrame(x, columns=data.feature_names)
print("\nFirst five rows:")
print(df.head())

print("\nNumber of samples:", x.shape[0])
print("Number of features:", x.shape[1])
print("Number of classes:", len(np.unique(y)))
print("Class names:", data.target_names)
```

## Task 3: Split the Dataset
Split the dataset into:
* 80% training
* 20% testing


```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

print("\n=== Task 3: Split the Dataset ===")
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## Task 4: Standardize Features
Apply feature scaling using StandardScaler as you did in previous lab (Lab 2).
(Note: import StandardScaler from sklearn.preprocessing)

```
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Task 5: Train Logistic Regression Model
Train the logistic regression model using default parameters.
(Note: import LogisticRegression from sklearn.linear_model)

```
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)
```

## Task 6: Make Predictions
Predict class labels for the test set.

```
y_pred = model.predict(X_test_scaled)
```

## Task 7: Evaluate the Model
Compute following parameters to evaluate your model:
1. Accuracy
2. Precision
3. Recall
4. F1-score
5. Confusion Matrix
(Note: import accuracy_score from sklearn.metrics)

```
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("\n=== Task 7: Evaluate the Model ===")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")
print("Confusion Matrix:")
print(cm)
```

## Task 8: Interpret Coefficients
Display the coefficients of the trained model.
Questions:
1. Which feature has the largest positive coefficient?
2. Which feature has the largest negative coefficient?
3. What does the sign of a coefficient indicate?

```
coefficients = pd.Series(model.coef_[0], index=data.feature_names)
coefficients_sorted = coefficients.sort_values()

print("\n=== Task 8: Interpret Coefficients ===")
print(coefficients_sorted)

largest_positive = coefficients.idxmax()
largest_negative = coefficients.idxmin()

print(f"\nFeature with largest positive coefficient: {largest_positive} "
      f"({coefficients[largest_positive]:.4f})")
print(f"Feature with largest negative coefficient: {largest_negative} "
      f"({coefficients[largest_negative]:.4f})")
```
